```python
import pandas as pd

data = {
    "Date": [
        "2024-01-05","2024-01-10","2024-02-02","2024-02-15",
        "2024-03-01","2024-03-12","2024-03-20","2024-04-05",
        "2024-04-18","2024-05-02"
    ],
    "Product": [
        "Shoes","Shirt","Shoes","Hat",
        "Shirt","Shoes","Hat","Shirt",
        "Shoes","Hat"
    ],
    "Region": [
        "North","South","East","West",
        "North","East","South","West",
        "North","South"
    ],
    "Quantity": [10,5,8,3,7,12,4,6,9,2],
    "Price": [700,300,700,200,300,700,200,300,700,200]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
df
```

```python
df["Revenue"].sum()
```

```python
df.groupby("Product")["Revenue"].sum()
```

```python
df.groupby("Region")["Revenue"].sum()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Month"] = df["Date"].dt.month
```

```python
df.groupby("Month")["Revenue"].sum()
```

```python
import matplotlib.pyplot as plt

monthly_sales = df.groupby("Month")["Revenue"].sum()

monthly_sales.plot(kind="bar")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
import matplotlib.pyplot as plt

monthly_sales = df.groupby("Month")["Revenue"].sum()

plt.figure(figsize=(9,5))

plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title("Monthly Revenue Trend", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(monthly_sales.index)
plt.grid(True)

plt.tight_layout()

# SAVE IMAGE (important line)
plt.savefig("../visuals/monthly_revenue_trend.png")

plt.show()
```

## Key Business Insights

### 1. Product Revenue Concentration
Shoes contribute the largest portion of total revenue, significantly outperforming other products. This indicates that revenue is heavily dependent on a single product category. While this creates strong sales performance, it also introduces business risk if demand declines. Diversifying promotions across other products could stabilize revenue streams.

### 2. High-Value vs Low-Volume Products
Although some products sell in smaller quantities, higher pricing contributes significantly to revenue generation. This suggests that pricing strategy plays a major role in profitability, and maintaining premium pricing for high-demand items may maximize earnings.

### 3. Regional Sales Opportunity
The North region consistently generates higher revenue compared to other regions, suggesting stronger market demand or customer concentration. Expanding marketing campaigns or improving distribution channels in this region could further increase sales performance.

### 4. Seasonal Sales Pattern
Revenue peaks around March, indicating potential seasonal purchasing behavior. Preparing inventory and promotional campaigns ahead of peak months could prevent stock shortages and capture additional revenue opportunities.

### 5. Revenue Stability Observation
Monthly revenue shows fluctuations after peak periods, which may indicate inconsistent customer purchasing patterns. Implementing loyalty programs or recurring promotions could help stabilize monthly income.

### 6. Strategic Business Recommendation
The company should prioritize inventory planning for high-performing products, invest in strong regions, and introduce targeted promotions during slower months to maintain consistent revenue growth.
